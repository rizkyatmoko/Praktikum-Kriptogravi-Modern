```python
def vigenere_encrypt(plaintext, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in plaintext:
        if char.isalpha():
            shift = ord('A')
            char = char.upper()
            k = ord(key[key_index % len(key)]) - shift
            result += chr((ord(char) - shift + k) % 26 + shift)
            key_index += 1
        else:
            result += char

    return result


def vigenere_decrypt(ciphertext, key):
    result = ""
    key = key.upper()
    key_index = 0

    for char in ciphertext:
        if char.isalpha():
            shift = ord('A')
            char = char.upper()
            k = ord(key[key_index % len(key)]) - shift
            result += chr((ord(char) - shift - k) % 26 + shift)
            key_index += 1
        else:
            result += char

    return result


pesan = "KRIPTOGRAFI"
kunci = "KEYP"

cipher = vigenere_encrypt(pesan, kunci)
plain = vigenere_decrypt(cipher, kunci)

print("Ciphertext:", cipher)
print("Decrypted :", plain)
```
